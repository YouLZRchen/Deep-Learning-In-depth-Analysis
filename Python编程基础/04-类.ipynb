**类**是Python中非常重要的一个概念，它是一种抽象，其实也可以理解为“模板”，用来描述具有相同属性和方法的对象集合。类是对象的蓝图，它定义了对象的属性和行为。在Python中，类是通过关键字`class`来定义的。

### 1.类的定义

在创建类前，我们先明确一些定义：

- **类**：类是一个抽象的概念，它定义了一组具有相同属性和方法的对象的蓝图。
- **对象**：对象是类的实例，它具有类的属性和方法。
- **属性**：属性是对象的特征，它描述了对象的状态。
- **方法**：方法是对象的动作，它描述了对象的行为。
- **类属性**：类属性是定义在类中的变量，为所有实例对象共享。
- **实例属性**：实例属性是定义在类中的变量，为每个实例对象独有。
- **类方法**：类方法是定义在类中的方法，使用`@classmethod`装饰器进行修饰，第一个参数是类本身。
- **实例方法**：实例方法是定义在类中的方法，使用`self`参数进行修饰，第一个参数是实例对象。
- **静态方法**：静态方法是定义在类中的方法，使用`@staticmethod`装饰器进行修饰，没有参数。

```python
class Student:
    school='康杰中学'  #这就是个类属性：定义在类中，方法外的变量
    #初始化方法
    def __init__(self,name,age):#self代表实例对象,name和age是方法的参数，是局部变量，其作用域是整个init方法，不能在类外使用
        self.name=name#self.name是实例属性，name是局部变量
        self.age=age#self.age是实例属性，age是局部变量
    #定义在类中的函数，称为方法，自带一个参数self
    def show(self):
        print('我叫{0},今年{1}岁了'.format(self.name,self.age))
    #静态方法
    @staticmethod
    def sm():
        #print(self.name),报错
        print('这是一个静态方法，不能调用实例方法，也不能调用实例属性')
    #类方法
    @classmethod
    def cm(c):
        #print(self.name),报错
        print('这是一个类方法，不能调用实例方法，也不能调用实例属性')

#创建类的对象
stu1=Student('lzr',18)
#实例属性，使用对象名进行打点调用
print(stu1.name,stu1.age)
#类属性，直接使用类名打点调用
print(Student.school)
#实例方法，使用对象名进行打点调用
stu1.show()
#类方法，@classmethod进行修饰的方法，直接使用类名打点调用
Student.cm()
#静态方法，@staticmethod进行修饰的方法，直接使用类名打点调用
Student.sm()
```

### 2.类的动态绑定属性和方法

在Python中，类的属性和方法可以在运行时动态绑定，这意味着可以在创建对象之后，再为对象添加属性和方法。

```python
class Student:
    school='康杰中学'
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):
        print('我叫{0},今年{1}岁了'.format(self.name,self.age))
stu1=Student('lzr',18)
stu2=Student('cxz',18)
#为cxz动态绑定一个属性
stu2.gender='男'
print(stu2.name,stu2.gender,stu2.age)
#动态绑定方法
def introduce():
    print('我是一个普通的函数，我被动态绑定成了stu2的方法')
stu2.woshi=introduce#函数赋值，不能加小括号，否则为调用
stu2.woshi()#调用方法
```

### 3.类的继承

在Python中，类可以继承其他类的属性和方法。继承是面向对象编程中的一个重要概念，它允许我们创建一个新类，该类继承自另一个类，并可以添加新的属性和方法，或者重写父类的属性和方法。

```python
class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):
        print(f'大家好，我是{self.name}，今年{self.age}岁')

class Student(Person):#表示继承父类，Student是子类，Person是父类
    def __init__(self,name,age,stuno):#初始化
       super().__init__(name,age)#调用父类的属性
       self.stuno=stuno

class Doctor(Person):#一个父类可以拥有多个子类
    def __init__(self,name,age,department):
        super().__init__(name,age)
        self.department=department

stu=Student('lzr',18,950)
stu.show()
doc=Doctor('cxz',17,'神经外科')
doc.show()


```

### 4.类的多继承

在Python中，一个类可以继承多个父类，这种继承方式称为多继承。多继承允许一个类拥有多个父类的属性和方法，某些情况下非常有用。

```python
class FatherA():
    def __init__(self,name):
        self.name=name
    
    def showA(self):
        print('父类A中的方法')

class FatherB():
    def __init__(self,age):
        self.age=age
    
    def showB(self):
        print('父类B中的方法')

class Son(FatherA,FatherB):#多继承
    def __init__(self,name,age,gender):
        #需要调用两个父类的初始化方法
        FatherA.__init__(self,name)
        FatherB.__init__(self,age)
        self.gender=gender
    
son=Son('wqd',12,'男')
son.showA()
son.showB()
```

### 5.类的权限控制

在Python中，类的属性和方法可以通过权限控制来限制访问。权限控制是通过在属性和方法名前添加下划线或双下划线来实现的。下划线表示受保护的属性或方法，双下划线表示私有的属性或方法。

```python
class Student():
    def __init__(self,name,age,gender):
        self._name=name #self._name受保护，只能本类和子类访问
        self.__age=age #self.__age表示私有的，只能类本身去访问
        self.gender=gender #普通的实例属性，类的内部，外部，子类都可以访问
    
    def _fun1(self):
        print("子类及本身可以访问")

    def __fun2(self):
        print('只能定义的类去访问')

    def show(self):
        self._fun1() #类本身访问受保护的方法
        self.__fun2() #类本身访问私有方法
        print(self._name)
        print(self.__age)

#创建一个学生类的对象
stu=Student('lzr',18,'男')
#类的外部
print(stu._fun1())
#print(stu.__fun2())，报错，私有方法

#如何访问私有属性及方法
print(stu._Student__age)
stu._Student__fun2()
```

### 6.类中属性的设置

即面向对象编程中的封装以及如何使用@property装饰器来控制对私有属性的访问和修改。

```python
class Student():
    def __init__(self,name,gender):
        self.name=name
        self.__gender=gender #私有实例属性
    
    #使用@property方法，将方法转为属性使用
    @property
    def gender(self):
        return self.__gender
    #将gender这个属性改为可写属性
    @gender.setter #使用属性名称
    def gender(self,value):
        if value!='男' and value!='女':
            print('输入有误，已默认设置为男')
            self.__gender='男'
        else:
            self.__gender=value

    
stu=Student('lzr','男')
print(stu.name,stu.gender)

#尝试修改属性值
stu.gender='其他'
print(stu.name,stu.gender)
```

### 7.类的深拷贝与浅拷贝

在Python中，对象的拷贝分为浅拷贝和深拷贝。浅拷贝只复制对象本身，不复制对象内部的子对象，而深拷贝则复制对象本身及其内部的子对象。

```python
class CPU():
    pass
class Disk():
    pass

class Computer():
    #计算机由CPU和硬盘组成
    def __init__(self,cpu,disk):
        self.cpu=cpu
        self.disk=disk

cpu=CPU()
disk=Disk()
#创建一个计算机对象
com=Computer(cpu,disk)
#变量的赋值
com1=com
print(com,'子对象的内存地址:',com.cpu,com.disk)
print(com1,'子对象的内存地址:',com1.cpu,com1.disk)
print('-'*40)
#类对象的浅拷贝
import copy
com2=copy.copy(com)
print(com,'子对象的内存地址:',com.cpu,com.disk)
print(com2,'子对象的内存地址:',com2.cpu,com2.disk)
#类对象的深拷贝
print('-'*40)
com3=copy.deepcopy(com)
print(com,'子对象的内存地址:',com.cpu,com.disk)
print(com3,'子对象的内存地址:',com3.cpu,com3.disk)
```

### 8.方法重写

在Python中，子类可以重写父类的方法，即子类可以定义一个与父类方法同名的方法，从而覆盖父类的方法。重写方法可以改变子类的行为，以满足子类的特定需求。

```python
class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):
        print(f'大家好，我是{self.name}，今年{self.age}岁')

class Student(Person):
    def __init__(self,name,age,stuno):
       super().__init__(name,age)
       self.stuno=stuno
    #方法重写部分
    def show(self):
        #调用父类当中的方法
        super().show()
        #重新编写方法体
        print('我来自康杰中学，我的学号是{0}'.format(self.stuno))
    def __str__(self):
        return 'helloworld'


class Doctor(Person):
    def __init__(self,name,age,department):
        super().__init__(name,age)
        self.department=department
    #完全重写
    def show(self):
        print('大家好，我叫{0}，今年{1}岁，在{2}科室工作'.format(self.name,self.age,self.department))

    

stu=Student('lzr',18,950)
stu.show()
doc=Doctor('cxz',17,'神经外科')
doc.show()#调用子类自己的show方法

#__str__方法重写
print(stu)#<__main__.Student object at 0x000002347B27BF50>
#改写后输出helloworld
```

### 9.特殊方法

在Python中，特殊方法（也称为魔术方法）是类中的一些特殊方法，它们以双下划线开头和结尾，用于实现一些特殊的功能。例如，`__init__`方法用于初始化对象，`__str__`方法用于返回对象的字符串表示形式，`__len__`方法用于返回对象的长度等。

```python
class strange:
    def __init__(self,name):
        self.name = name
    def __add__(self, other):
        return self.name + other.name+'1'

if __name__ == '__main__':
    a = strange('a')
    b = strange('b')
    print(a+b)
```
