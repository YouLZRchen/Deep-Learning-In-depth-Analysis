**函数**是Python中非常重要的一个概念，它允许我们将一段代码封装起来，以便在需要的时候调用。函数可以接受参数，并返回结果。在Python中，函数的定义和使用都非常简单。

### PART1.可迭代对象

在Python中，可迭代对象是指可以逐一获取其元素的对象。常见的可迭代对象包括列表、元组、字符串、字典、集合和文件等。我们可以使用for循环来遍历可迭代对象中的元素。这一部分前面已经提及，此处给他们一个统一的名称。

### PART2.函数

**1.1 函数的定义与调用**

函数是一段可重复使用的代码块，它接受输入参数并返回结果。函数的定义使用关键字`def`，后跟函数名和参数列表，最后是函数体。函数体由缩进的代码块组成，用于执行特定的任务。首先展示一些我们常见的函数：

```python
print(min([1,2,3,4,5,6]))
print(sum([10,20,30]))
print(round(3.1415926))
```

函数的定义需要def来实现，函数名可以自定义，参数列表可以自定义，函数体可以自定义，返回值可以自定义。调用函数只需要函数名加括号即可。

```python
#定义一个函数s
def s(a,b):
    s=0
    for i in range(a,b+1):
        s+=i
    print(s)
s(2,4)
s(10,100)
```

**1.2 函数的传参**

函数的传参有两种方式：位置传参和关键字传参。

```python
def happy_birthday(name,age):
    print(name+'生日快乐',end=' ')
    print(str(age)+'岁生日快乐')
    #定义生日函数
#位置传参
happy_birthday('lzr',18)
#happy_birthday(18,'lzr')，位置没有对应正确，数据类型对不上，报错
#关键字传参
happy_birthday(age=18,name='lzr')
#happy_birthday(age=18,name1='lzr')，这个会报错，原函数中无name1这个变量
happy_birthday('cxz',age=17)#正常执行
#happy_birthday(name='cxz',17)，会报错，原则：位置参数在前，关键词参数在后
```

实际在设计函数时，往往会设定一些默认参数，这样在调用函数时，如果没有传入对应的参数，函数就会使用默认值：

```python
def happy_birthday(name='lzr',age=18):
    print(name+'生日快乐',end=' ')
    print(str(age)+'岁生日快乐')#定义生日函数
happy_birthday()
happy_birthday('cxz')
happy_birthday(age=17)
```

也存在个数可变的位置参数和个数可变的关键词参数，这种参数在函数定义时，需要在参数名前加上*和**，这样在调用函数时，可以传入任意数量的位置参数或关键词参数：

```python
#个数可变的位置参数
def kobe(*man):
    print(type(man))
    for i in man:
        print(i)
kobe(10,20,30,40)
kobe(10)
kobe([11,22,33,45])
kobe(*[11,22,33,45])

'''
最后输出的类型都是元组，因为*man会将传入的参数打包成一个元组，而**out会将传入的参数打包成一个字典。
'''
```

```python
#个数可变的关键词参数
def kobe2(**out):
    print(type(out))
    for a,b in out.items():
        print(a,'---->',b)
kobe2(name='lzr',age=18,x=20)
dict={'name':'lzr','age':18,'x':20}
#kobe2(dict)，报错
kobe2(**dict)
```

**1.3 全局变量与局部变量**

在Python中，全局变量是指在整个程序中都可以访问的变量，而局部变量是指只能在函数内部访问的变量。全局变量在函数内部可以直接使用，但是如果在函数内部修改了全局变量的值，那么这个修改只会在函数内部生效，不会影响到函数外部的全局变量。如果想在函数内部修改全局变量的值，可以使用`global`关键字来声明这个变量是全局变量。

```python
#全局变量
x1=100
def calc2(x,y):
    return x1+x+y
print(calc2(10,20))
#局部变量
def calc3(x,y):
    x1=200
    return x1+x+y
print(calc3(10,20))#优先级：局部变量>全局变量

def calc4(x,y):
    global s  #s为局部变量，使用global后变为全局变量
    s=300
    return x+y+s
print(calc4(10,20))
print(calc4(s,20))
```
