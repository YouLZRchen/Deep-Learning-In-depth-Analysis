### PART1.字符串

**1.1** Python中，字符串是一种比较简单的数据类型，就从它开始吧。
首先定义一个简单的字符串word（字符串用单引号或双引号括起来）

下方两个单元格分别展示如何定义一个简单字符串以及对其的相关操作。实际上Python字符串存在大量操作，这里只展示一些简单的操作。

```python
word='HelloWorld'
```

```python
print(word)
print(word.lower())  #将字符串中的所有字母转换为小写
print(word.upper())  #将字符串中的所有字母转换为大写，不会改变word值
```

**1.2 字符串的分割**

```python
e_mail='lzr0814@163.com'
lst=e_mail.split('@')  #这里返回的是一个列表，关于列表会在后续介绍
print('邮箱名:',lst[0],'邮箱地址:',lst[1])
```

其他对字符串操作的函数

```python
#word='HelloWorld'
print(word.count('o'))#统计字符串中某个字符出现的次数
print(word.find('o'))#查找某个字符在字符串中的位置
print(word.find('t'))#会返回-1，因为字符串中没有这个字符
print(word.index('o'))#和find类似，但是找不到会报错
#判断前缀和后缀
print(word.endswith('ld'))
print(word.startswith('He'))
print(word.startswith('Hel'))
print(word.startswith('hel'))
```

**1.3 字符串的某个字符替换**

```python
new_word=word.replace('o','a',1)#替换字符串中的某个字符，第三个参数表示替换的次数，默认是全部替换
print(new_word)
```

**1.4 多个字符串的拼接**  

此处主要展示多个字符串拼接的数个方法，除下述三种方法外，还有格式化字符串的方法，后续会介绍。

```python
s1='hello'
s2='world'
#（1）使用'+'拼接
print(s1+s2)
#（2）使用join拼接，join函数本质就是将列表中的元素以某个字符连接起来，语法为：''.join([s1,s2])，其中''表示连接的字符
print(''.join([s1,s2]))
print('123'.join([s1,s2,'python']))
#（3）直接拼接
print('hello''world')
```

**1.5 字符串的格式化**

```python
name='张三'
age=18
score=98.5
#（1）使用占位符进行格式化
print('我叫%s,今年%d岁,成绩是%.1f'%(name,age,score))#%s表示字符串，%d表示整数，%.1f表示小数，1表示小数点后一位
#(2)f-string
print(f'我叫{name},今年{age}岁,成绩是{score}')#f-string是Python3.6以后才有的，使用f-string可以更方便地进行字符串格式化
#(3)使用字符串的format方法
print('姓名：{0},年龄：{1}，成绩：{2}'.format(name,age,score))
print('姓名：{2},年龄：{0}，成绩：{1}'.format(age,score,name))#注意对应关系
#格式化字符串的详细格式
S='helloworld'
print('{0:*<20}'.format(S))#字符串的长度为20，左对齐，空白位置用*填充
print('{0:*>20}'.format(S))#字符串的长度为20，右对齐，空白位置用*填充
print('{0:*^20}'.format(S))#字符串的长度为20，居中，空白位置用*填充
print('{0:,}'.format(1145149))
print('{0:,}'.format(1145149.3847))#三位一逗
print('{0:.2f}'.format(3.1415926))#控制浮点数精确度
print('{0:.4}'.format(S))
```

```python
#字符串在指定的宽度范围内居中
print(word.center(20,' '))#在20个字符的宽度内居中，不够用空格填充
```

**1.6 字符串去除指定字符**

此操作主要借助strip()函数实现，具体用法如下：

```python
#去掉字符串左右的空格
print('  Hello World  '.strip())#去掉字符串左右两边的空格
print('  Hello World  '.lstrip())#去掉字符串左边的空格
print('  Hello World  '.rstrip())#去掉字符串右边的空格
#去掉指定的字符
s='dl-HelloWorld'
print(s.strip('dl-'))#去掉字符串左右两边的指定字符，与顺序无关
print(s.lstrip('dl-'))#去掉字符串左边的指定字符
print(s.rstrip('dl-'))#去掉字符串右边的指定字符
```

**1.7 字符串的编码与解码**

计算机上的文件、内存、网络传输等都是以二进制的形式存储的，而字符串是以文本的形式存储的，因此需要将字符串转换为二进制，这个过程称为编码，反之称为解码。常见的编码方式有ASCII、UTF-8、GBK等。

```python
cn='伟大的中国梦'
#字符串的编码
scode=cn.encode(errors='replace')
print(scode)#默认是utf-8模式，一个英文字母是一个字节，一个汉字是三个字节
scode_gbk=cn.encode(encoding='gbk',errors='replace')
print(scode_gbk)#gbk模式中，一个汉字是两个字节
#字符串的解码
print(bytes.decode(scode))
print(bytes.decode(scode_gbk,'gbk'))
```

**1.8 字符串的切片**

有个很重要一点，Python中大部分数据类型正索引都是从0开始的，字符串也不例外。负索引默认从字符串的最后一个字符开始，-1表示最后一个字符，-2表示倒数第二个字符，以此类推。

```python
s='helloworld'
print(s[0:5:2])#索引从零开始，到五结束（不包含五），步长为2，索引默认为0，步长默认为1，结束索引默认为最后一个元素
print(s[::-1])#逆序输出
print(s[::-2])#逆序步长为2输出
```

字符串的max和min函数本身没有过多的意义，主要是输出字符串中ASCII码值最大的字符和最小的字符。

ASCII值（美国信息交换标准代码）是计算机中字符的编码，每个字符都有一个对应的ASCII值，比如大写字母A的ASCII值为65，小写字母a的ASCII值为97，数字0的ASCII值为48，空格的ASCII值为32，回车符的ASCII值为13，换行符的ASCII值为10，制表符的ASCII值为9。

```python
print(max(s))
print(min(s))
```
