列表，集合，字典也是Python中十分常用的数据类型，它们可以存储多个数据，并且可以方便地对其进行操作。此部分就简单地介绍这些数据类型的创建、访问和操作。

有一些常用的函数或操作如for,in,len,max,min等可以类比，有的在此处可能尚未展开

### PART1. 列表

**1.1 列表的创建**

列表数据类型中元素可以是任意类型，包括数字、字符串、列表、字典等。同时元素有序且可重复。

```python
lst=[1,2,3,4,5]
lst1=[i for i in range(10)]
print(lst1)
```

**1.2 列表元素的访问**

其实很多数据类型的操作基本雷同，索引从0开始，切片的格式为[start:end:step]

```python
lst=[1,2,3,4,5]
print(lst[0])
print(lst[1:3])
print(lst[::2])
```

**1.3 向列表添加元素**

```python
lst=[1,2,3,4,5]
lst.append(6)
print(lst) #添加单个元素

lst1=[7,8,9]
lst.extend(lst1)
print(lst) #通过另一列表添加多个元素
```

**1.4 列表删除元素**

```python
lst=[1,2,3,4,5]
del lst[0]
print(lst) #通过索引删除元素(也可采用pop函数)

lst.remove(3)
print(lst) #通过值删除元素（对于重复元素，只会删除第一个）

lst.clear()
print(lst) #清空列表
```

**1.5 列表的排序**

```python
lst=[3,4,1,2,5]
lst.sort()
print(lst) #升序排序

lst.sort(reverse=True)
print(lst) #降序排序

lst1=[1,2,3,4,5]
lst1.reverse()
print(lst1) #反转列表
```

### PART2. 集合

**2.1 集合的创建**

集合也是一种数据类型，但其内部元素是无序的，不重复的

```python
s1={1,2,3,4}
print("s1:",s1)#列表不能作为集合中的元素，会报错
s2=set()#set用于创建集合
print("s2:",s2)#s2是空集，如果写s={},print(s),此时s不是空集，而是一个字典
s3=set('helloworld')
print("s3:",s3)#集合是无序的，不重复的，故只会输出一个l，一个o
s4=set([10,20,30])
print("s4:",s4)
s5=set(range(1,10))
print("s5:",s5)

#生成式创建集合
S={item*2 for item in range(10,100) if item%5==1}
print(S)
```

**2.2 集合的交并补**

```python
A={1,2,3,4,5}
B={2,4,6,8}
print(A&B)#交集
print(A|B)#并集
print(A-B)#差集（左侧集合减去两个集合的公共元素）
print(A^B)#补集（二者的所有元素除去公共元素）
```

**2.3 集合的遍历**

```python
S2={1231,3422,354,4667,5675,6675}
for i in S2:
    print(i,end=' ')
for a,b in enumerate(S2):#编号并输出元素
    print(a,b)
```

**2.4 集合的增删改**

```python
S={1,2,3}
S.add(100)
print(S)
S.remove(2)
print(S)
S.clear()
print(S)
```

### PART3. 字典

**3.1 字典的创建**

字典是一种键值对的数据类型，键值对之间用冒号隔开，键值对之间用逗号隔开，整个字典用大括号括起来。键值对之间无序，键值对之间无重复。
d1={'name':'Tom','age':18,'gender':'male'}，元组可以作为字典中的键,列表不行！

```python
#字典创建方法1
d1={10:'cat',20:'dog',30:'pet',20:'zoo'}#大括号内为key:value
print(d1)#第二个20的zoo将第一个20的dog进行了一个覆盖
#字典创建方式2
lst1=[10,20,30,40]
lst2=['kobe','mamba','牢大']#创建两个列表
d_2=zip(lst1,lst2)
print(d_2)#无法看到内容，<zip object at 0x0000022AE656ED00>
d2=dict(d_2)
print(d2)#使用dict函数进行一个转化，输出{10: 'kobe', 20: 'mamba', 30: '牢大'}
#字典创建方式3
d3=dict(a=1,b=2,c=3)#等号左侧是key，右侧是value
print(d3)

#生成式创建字典
import random  #引入随机数
d5={k:random.randint(1,100) for k in range(4)}#1,100生成随机数作为值对应0,1,2,3组成字典
print(d5)
```

**3.2 字典元素的访问**

```python
d1={10:'cat',20:'dog',30:'pet',20:'zoo'}
print(d1[10])
print(d1.get(10))
#二者区别演示
#print(d1['man'])，会报错，注释掉
print(d1.get('man'))#不会报错，报出None(默认值)
print(d1.get('man','kobe'))#可以指定默认值，此处为kobe
```

**3.3 字典的遍历**

```python
d1={10:'cat',20:'dog',30:'pet',20:'zoo'}
for i in d1.items():                   
    print(i)#输出的是元组
for a,b in d1.items():
    print(a,'---->',b)
```

**3.4 字典的增删改**

```python
d1={10:'cat',20:'dog',30:'pet',20:'zoo'}
#向字典中添加元素
d1[200]='cow'
print(d1)#如果写d1[20]='cow'，会覆盖掉zoo，即改
```

```python
lst=list(d1.items())
print(lst)
#将列表转为字典
d4=dict(lst)
#将元素从字典中取出
print(d4.pop(20))
print(d4)
print(d4.pop(114514,'不存在'))
#随机删除
print(d4.popitem())
print(d4)
#清空字典中的所有元素
d4.clear()
print(d4)
print(bool(d4),bool(d1))
```

**3.5 字典键值的获取**

```python
d1={10:'cat',20:'dog',30:'pet',20:'zoo'}
#获取字典中所有的键（keys）
print(d1.keys())
print(list(d1.keys()))#列表形式
print(tuple(d1.keys()))#元组形式
#获取字典中所有的值（values）
print('-'*50)
print(d1.values())
print(list(d1.values()))#列表形式
print(tuple(d1.values()))#元组形式
```
