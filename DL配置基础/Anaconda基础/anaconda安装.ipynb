### 下载方法

anaconda当前可在下述链接（官网下载）
https://www.anaconda.com/download/success?reg=skipped

该界面主要包括anaconda与miniconda两种安装包，可自行选择下载，Miniconda，体积很小，只装 conda+python等，库自己按需装，不占大量磁盘空间。Anaconda 是全家桶，预装几百个包，占用 3GB 以上磁盘。

安装在磁盘后即可开始使用

### 安装后配置

anaconda的许多操作可在终端实现

如要在Windows cmd运行，需提前将其配置在环境变量，即以下三条路径：

"E:\miniconda3"

"E:\miniconda3\Library\bin"

"E:\miniconda3\Scripts"

当然，也可打开Anaconda Prompt后运行相关指令
