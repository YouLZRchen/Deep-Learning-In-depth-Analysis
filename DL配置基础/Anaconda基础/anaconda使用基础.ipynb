### PART1.anaconda环境创建

使用anaconda可以在cmd中使用指令实现

**查看anaconda版本与python版本：**

conda --version

python -V

**查看当前所有环境**

conda env list

**创建新环境**（--name后为环境名称，名称后为py版本）

conda create --name py38env python=3.8

之后可能会询问是否接受一堆ToS协议，接受即可，或者通过下述指令每次创建时默认接受：

conda config --set repo_accept_terms true

**创建环境后激活环境**

activate py38env

之后命令行前面的括号内即为当前环境名称

**删除环境**

conda remove -n 环境名 --all

### PART2.对环境操作

查看当前环境安装的所有包：

conda list

**安装新包**

conda install numpy

**退出当前环境**

conda deactivate

### PART3.VScode中采用anaconda相关操作

Ctrl+Shift+P后点击选择解释器后展示列表，可以选择命令行创建的环境，也可在此处自行创建环境

创建虚拟环境 --> Conda --> 创建命名conda环境（每个项目都能找到，若选择前缀，则仅当前项目能看到） --> 选择Python环境
