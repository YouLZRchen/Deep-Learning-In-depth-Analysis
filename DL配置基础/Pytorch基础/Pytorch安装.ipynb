安装pytorch前，可以先看一下自身电脑是否带有显卡

右击开始 --> 设备管理器 --> 显示适配器 --> 是否有NVIDIA RTX等显卡

**先创建一个conda虚拟环境**（参考anaconda基础，以下仅为例子）

conda create --name torchmodel python=3.12

conda activate torchmodel

接下来打开链接https://pytorch.org，并翻到Install PyTorch部分，根据个人需求选择需要的内容再复制下方Run this command内容即可（最新稳定版 PyTorch，Python 版本必须 ≥3.10，下述为示例指令，复制到激活环境后的命令行即可）

pip3 install torch torchvision --index-url https://download.pytorch.org/whl/cu126

安装完毕使用该环境做项目即可


**附：** 几种CUDA的区别

这里是PyTorch 编译时绑定的 CUDA 版本，并非CUDA Toolkit。

NVIDIA 驱动有最低版本要求，你的显卡驱动必须 ≥ 这个版本才能跑对应的 PyTorch‑CUDA：

CUDA 12.6 → 驱动 ≥ 560.28.03

CUDA 13.0 → 驱动 ≥ 570.86.12

CUDA 13.2 → 驱动 ≥ 580.24.03

**Windows查看驱动：** NVIDIA 控制面板 → 系统信息 → 驱动版本。

CUDA 12.6兼容性最好，绝大多数 50系、40系、30系显卡驱动都支持。

CUDA13.0、13.2只支持比较新的显卡驱动。对 RTX50 系做专门优化，少量算子速度提升。属于比较新的版本，偶尔会遇到少量第三方库不兼容。
